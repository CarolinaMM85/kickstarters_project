## Data Quality Assessment

```python
# Most data quality issues were concentrated in the Name field,
# reflecting the variability of user-generated content rather than
# structural problems in the dataset.
```

```python
#the ~ operator in Pandas is simply the “NOT” operator. It flips a boolean mask.
```

```python
# Base column
names = kickstarter_data["Name"]

# Conditions
has_profanity = names.str.contains(r"F\$#%", case=False, na=False)
has_encoding_issue = names.str.contains(r"â€”|Â|â€", na=False)   # common broken encodings
has_status_tag = names.str.contains(r"\(Canceled\)|\(Lostwars\)", na=False)
has_excessive_punct = names.str.contains(r"[!?]{2,}", na=False)  # two or more ! or ?
has_symbols = names.str.contains(r"[@#$%^&*]", na=False) # stray symbols
has_bad_case = names.str.contains(r"^limited Edition$", case=False, na=False) #only rows where the entire title is “limited Edition” will be flagged
has_hashtags = names.str.contains(r"#MEGA", case=False, na=False)
has_all_caps = names.str.contains(r"Americas Got Talent - Serious MAK", case=False, na=False)
has_marketing_exaggerations = names.str.contains(r"Forge in Your Own Fire", case=False, na=False)
has_special_formatting = names.str.contains(r"eVo Planner|o2f", na=False)
has_product_bundles = names.str.contains(r"\band\b|\bplus\b|\bwith\b", na=False)
# Marks titles that include the separate words "and", "plus", or "with". 
# These words usually mean the project is offering more than one item together (a bundle). 
# The \b makes sure it only matches the whole word, not inside other words. 
# na=False just skips over empty cells instead of breaking.

# Combine with OR (any of these issues)
needs_cleaning = (
    has_profanity |
    has_encoding_issue |
    has_status_tag |
    has_excessive_punct |
    has_symbols |
    has_bad_case |
    has_hashtags |
    has_all_caps |
    has_marketing_exaggerations |
    has_special_formatting |
    has_product_bundles
)


# Keep only rows that need cleaning, Name column only
special_rows = kickstarter_data[needs_cleaning][["Name"]]
```

```python
#numeric_suffixes
kickstarter_data.Name.replace(to_replace=r'(\d+\.\d+)', regex=True, value= r'V\1')
kickstarter_data["Name"]

# Regex replacement: find version numbers with a decimal (e.g. 2.0, 3.14) in the Name column.
# Pattern (\d+\.\d+) captures both sides of the decimal as group 1.
# Replacement r'V\1' prefixes the captured version with 'V' → "Project 2.0" becomes "Project V2.0".
# Using r'' ensures backslashes are treated literally in the regex.
```

```python
#Encoding / spacing issues
kickstarter_data['Name'] = kickstarter_data['Name'].str.replace(r':\s+', ':')
kickstarter_data["Name"]
```

```python
#Capitalize the the first letter of each string in Name
kickstarter_data['Name'] = kickstarter_data['Name'].str.title() 
kickstarter_data["Name"]

special_rows = kickstarter_data[needs_cleaning]
special_rows.to_csv(r"C:\Users\Carolina_Muñoz\Desktop\Kickstarter Campaign Analytics\data\special_rows.csv", index=False)

```

## Data Reload Validation and Dataset Integrity Verification

```python
# Validation confirmed that transformations were successfully applied
# without introducing unexpected errors or data loss.
# All campaign records were preserved during preprocessing, ensuring
# that subsequent analyses are based on the complete Kickstarter dataset.
```

```python
import pandas as pd
import mysql.connector

# 1. Connect to MySQL
conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password="Faithnomore91.",   # replace with your real root password
    database="kickstarter_db"
)
cursor = conn.cursor()

# 2. Load CSV in chunks of 50,000
for chunk in pd.read_csv(
    r"C:\Users\Carolina_Muñoz\Desktop\Kickstarter Campaign Analytics\data\kickstarter_clean_fixed.csv",
    chunksize=50000
):
    rows = []
    for _, row in chunk.iterrows():
        rows.append((
            row['ID'], row['Name'], row['Category'], row['Subcategory'], row['Country'],
            row['Launched'], row['Deadline'], row['Goal'], row['Pledged'], row['Backers'], row['State']
        ))

    # Use execute many for efficiency
    cursor.executemany(
        """
        INSERT INTO campaigns (ID, Name, Category, Subcategory, Country, Launched, Deadline, Goal, Pledged, Backers, State)
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
        """,
        rows
    )
    conn.commit()
    print(f"Inserted {len(rows)} rows")

cursor.close()
conn.close()
```
